```python
'''
In pandas, standard NumPy data types (like int64 or float64) do not native-ly support missing values (NaN, None, <NA>), 
so trying to cast a column containing nulls into a strict standard type often raises an error
Pandas provides capitalized Extension Dtypes (e.g., Int64, Float64, boolean) designed 
specifically to hold valid data alongside missing values without throwing errors.
'''
for i in data:
    if data[i].dtype==float:
        data[i]=data[i].astype('Int64') 
data.dtypes
```

```python
encode=[i for i in data if (data[i].dtype!=int and data[i].dtype!='Int64')]
encode
```

```python
from sklearn.preprocessing import LabelEncoder
encoder=LabelEncoder()
for i in encode:
    #training_data[i]=encoder.fit_transform(training_data[i]) 
    data[i] = data[i].map(lambda x: encoder.fit_transform([x])[0] if type(x)==str else x)
data
```

Model Fitting

```python
h1n1_cols=[i for i in data if 'seas' not in i]
h1n1_data=data[h1n1_cols]
seas_cols=[i for i in data if 'h1n1' not in i]
seas_data=data[seas_cols]
seas_data
```

```python
from sklearn.model_selection import train_test_split
train_x1,test_x1,train_y1,test_y1 = train_test_split(data.drop('h1n1_vaccine',axis=1),data['h1n1_vaccine'],test_size=0.15,random_state=69,stratify=data['h1n1_vaccine'])
```

```python
train_y1.value_counts() #Cant use SMOTE with null values so just roll with this rn
```

```python
from sklearn.ensemble import RandomForestClassifier
model1 = RandomForestClassifier(class_weight='balanced')
model1.fit(train_x1,train_y1)
model1.score(test_x1,test_y1)
```

```python
from sklearn.metrics import classification_report
pred=model1.predict(test_x1)
pd.DataFrame(classification_report(test_y1,pred,output_dict=True))
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(test_y1,model1.predict(test_x1))
plt.figure(figsize=(6,6))
sns.heatmap(cm, annot=True, fmt=".3f", linewidths=.5, square = True, cmap = 'Blues_r')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
all_sample_title = 'Accuracy Score: {0}'.format(model1.score(test_x1,test_y1))
plt.title(all_sample_title, size = 15)
```

```python
from sklearn.model_selection import RandomizedSearchCV
params = {
    'max_depth': ([int(x) for x in np.linspace(10, 110, num=11)]+[None]), 
    'min_samples_split': [i*5 for i in range(1,30)],
    'min_samples_leaf': [i*3 for i in range(1,11)], 
    'bootstrap': [True, False]
}
model1_finder=RandomizedSearchCV(RandomForestClassifier(class_weight='balanced'),params,n_iter=200,cv=4, verbose=1, random_state=42,n_jobs=-1,scoring='f1')
model1_finder.fit(train_x1,train_y1)
```

```python
better_params=model1_finder.best_params_
better_params
```

```python
model1=RandomForestClassifier(**better_params,class_weight='balanced')
model1.fit(train_x1,train_y1)
print(model1.score(train_x1,train_y1),model1.score(test_x1,test_y1))
```

```python
from sklearn.metrics import classification_report
pred=model1.predict(test_x1)
pd.DataFrame(classification_report(test_y1,pred,output_dict=True))
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(test_y1,model1.predict(test_x1))
plt.figure(figsize=(6,6))
sns.heatmap(cm, annot=True, fmt=".3f", linewidths=.5, square = True, cmap = 'Blues_r')
plt.ylabel('Actual label')
plt.xlabel('Predicted label')
all_sample_title = 'Accuracy Score: {0}'.format(model1.score(test_x1,test_y1))
plt.title(all_sample_title, size = 15)
```

```python
from sklearn.metrics import f1_score
f1_score(test_y1,model1.predict(test_x1))
```
